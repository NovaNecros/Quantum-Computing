```python
from qiskit import QuantumRegister, ClassicalRegister, QuantumCircuit
from numpy import pi
```

```python
# | Phi+ >

qreg_q = QuantumRegister(2, 'q')
creg_c = ClassicalRegister(3, 'c')
circuit = QuantumCircuit(qreg_q, creg_c)

circuit.h(qreg_q[0])
circuit.cx(qreg_q[0], qreg_q[1])

circuit.draw()
```

```python
# Estado para teletransportar

def estado(circuit, i):
  circuit.x(qreg_q[i])
  circuit.h(qreg_q[i])
```

```python
# Teletransportación

qreg_q = QuantumRegister(3, 'q')
creg_c = ClassicalRegister(2, 'c')
circuit = QuantumCircuit(qreg_q, creg_c)

# Estado
estado(circuit, 0)

circuit.h(qreg_q[2])
circuit.cx(qreg_q[2], qreg_q[1])
circuit.cx(qreg_q[0], qreg_q[1])
circuit.h(qreg_q[0])
circuit.measure(qreg_q[0], creg_c[0])
circuit.measure(qreg_q[1], creg_c[1])
circuit.cx(qreg_q[1], qreg_q[2])
circuit.cz(qreg_q[0], qreg_q[2])

circuit.draw()
```

```python
# Deutsch

qreg_q = QuantumRegister(2, 'q')
creg_c = ClassicalRegister(2, 'c')
circuit = QuantumCircuit(qreg_q, creg_c)

circuit.x(qreg_q[1])
circuit.h(qreg_q[0])
circuit.h(qreg_q[1])

#oracle
circuit.cx(qreg_q[0], qreg_q[1])

circuit.h(qreg_q[0])
circuit.measure(qreg_q[0], creg_c[0])

circuit.draw()
```

```python
def oraculoGrover(n, circuit):
  if(n == 0):
    circuit.h(qreg_q[1])
    circuit.ccx(qreg_q[2], qreg_q[0], qreg_q[1])
    circuit.x(qreg_q[0])
    circuit.x(qreg_q[1])
    circuit.x(qreg_q[2])
    circuit.h(qreg_q[1])
  elif(n==1):
    circuit.h(qreg_q[0])
    circuit.x(qreg_q[1])
    circuit.x(qreg_q[2])
    circuit.ccx(qreg_q[1], qreg_q[2], qreg_q[0])
    circuit.h(qreg_q[0])
    circuit.x(qreg_q[1])
    circuit.x(qreg_q[2])
  elif(n==2):
    circuit.x(qreg_q[0])
    circuit.h(qreg_q[1])
    circuit.x(qreg_q[2])
    circuit.ccx(qreg_q[0], qreg_q[2], qreg_q[1])
    circuit.x(qreg_q[0])
    circuit.h(qreg_q[1])
    circuit.x(qreg_q[2])
  elif(n==3):
    circuit.h(qreg_q[0])
    circuit.x(qreg_q[2])
    circuit.ccx(qreg_q[1], qreg_q[2], qreg_q[0])
    circuit.h(qreg_q[0])
    circuit.x(qreg_q[2])
  elif(n==4):
    circuit.x(qreg_q[0])
    circuit.h(qreg_q[1])
    circuit.x(qreg_q[2])
    circuit.ccx(qreg_q[0], qreg_q[2], qreg_q[1])
    circuit.x(qreg_q[0])
    circuit.h(qreg_q[1])
    circuit.x(qreg_q[2])
  elif(n==5):
    circuit.h(qreg_q[0])
    circuit.x(qreg_q[2])
    circuit.ccx(qreg_q[1], qreg_q[2], qreg_q[0])
    circuit.h(qreg_q[0])
    circuit.x(qreg_q[2])
  elif(n==6):
    circuit.x(qreg_q[0])
    circuit.h(qreg_q[1])
    circuit.ccx(qreg_q[0], qreg_q[2], qreg_q[1])
    circuit.x(qreg_q[0])
    circuit.h(qreg_q[1])
  elif(n==7):
    circuit.h(qreg_q[0])
    circuit.ccx(qreg_q[1], qreg_q[2], qreg_q[0])
    circuit.h(qreg_q[0])
```

```python
#Grover

qreg_q = QuantumRegister(3, 'q')
creg_c = ClassicalRegister(3, 'c')
circuit = QuantumCircuit(qreg_q, creg_c)

circuit.h(qreg_q[0])
circuit.h(qreg_q[1])
circuit.h(qreg_q[2])

#Oráculo
oraculoGrover(5, circuit)

#Difusor
circuit.h(qreg_q[0])
circuit.h(qreg_q[1])
circuit.h(qreg_q[2])
circuit.x(qreg_q[0])
circuit.x(qreg_q[1])
circuit.x(qreg_q[2])
circuit.h(qreg_q[0])
circuit.ccx(qreg_q[1], qreg_q[2], qreg_q[0])
circuit.h(qreg_q[0])
circuit.x(qreg_q[0])
circuit.x(qreg_q[1])
circuit.x(qreg_q[2])
circuit.h(qreg_q[0])
circuit.h(qreg_q[1])
circuit.h(qreg_q[2])

circuit.measure(qreg_q[0], creg_c[0])
circuit.measure(qreg_q[1], creg_c[1])
circuit.measure(qreg_q[2], creg_c[2])

circuit.draw()
```
